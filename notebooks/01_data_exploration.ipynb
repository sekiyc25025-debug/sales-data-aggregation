```python
import pandas as pd
import re

def convert_reiwa_date(text):
    match = re.match(r"令和(\d+)年(\d+)月(\d+)日", text)

    reiwa_year = int(match.group(1))
    month = int(match.group(2))
    day = int(match.group(3))

    year = reiwa_year + 2018

    return pd.Timestamp(year=year, month=month, day=day)

df_a1 = pd.read_excel("../input/発注書A①.xlsx")
df_a2 = pd.read_excel("../input/発注書A②.xlsx")
df_b = pd.read_csv("../input/発注書B①.csv")

df_a_std1 = df_a1.rename(columns={
    "発注日": "order_date",
    "商品名": "product_name",
    "数量": "quantity"
})
df_a_std2 = df_a2.rename(columns={
    "発注日": "order_date",
    "商品名": "product_name",
    "数量": "quantity"
})
df_b_std = df_b.rename(columns={
    "発注日付": "order_date",
    "製品名": "product_name",
    "個数": "quantity"
})

df_b_std["order_date"] = df_b_std["order_date"].apply(convert_reiwa_date)

df_all = pd.concat(
    [df_a_std1, df_a_std2, df_b_std],
    ignore_index=True
)

df_summary = (
    df_all
    .groupby(["order_date", "product_name"], as_index=False)["quantity"]
    .sum()
)
```

```python
import pandas as pd
import re
from openpyxl import load_workbook

report_path = "../input/製品別売上レポート.xlsx"
output_path = "../output/製品別売上レポート.xlsx"

def convert_reiwa_date(text):
    match = re.match(r"令和(\d+)年(\d+)月(\d+)日", text)

    reiwa_year = int(match.group(1))
    month = int(match.group(2))
    day = int(match.group(3))

    year = reiwa_year + 2018

    return pd.Timestamp(year=year, month=month, day=day)

a_files = [
    "../input/発注書A①.xlsx",
    "../input/発注書A②.xlsx",
]
b_files = [
    "../input/発注書B①.csv",
    "../input/発注書B②.csv",
    "../input/発注書B③.csv"
]

a_dfs = []
b_dfs = []

for file in a_files:
    a_df = pd.read_excel(file)
    a_df = a_df.rename(columns={
        "発注日": "order_date",
        "商品名": "product_name",
        "数量": "quantity"
    })
    a_dfs.append(a_df)
    
for file in b_files:
    b_df = pd.read_csv(file)
    b_df = b_df.rename(columns={
        "発注日付": "order_date",
        "製品名": "product_name",
        "個数": "quantity"
    })
    b_df["order_date"] = b_df["order_date"].apply(convert_reiwa_date)

    b_dfs.append(b_df)

df_all = pd.concat(a_dfs + b_dfs,ignore_index=True)

df_summary = (
    df_all
    .groupby(["order_date", "product_name"], as_index=False)["quantity"]
    .sum()
)

wb = load_workbook(report_path)
ws = wb.active

for row_index, row in enumerate(df_summary.itertuples(index=False), start=2):
    date_cell = ws.cell(row=row_index, column=1, value=row.order_date)
    date_cell.number_format = "yyyy/mm/dd"

    ws.cell(row=row_index, column=2, value=row.product_name)
    ws.cell(row=row_index, column=3, value=row.quantity)

wb.save(output_path)
```

```python
import pandas as pd
import re
from copy import copy
from openpyxl import load_workbook

report_path = "../input/製品別売上レポート.xlsx"
output_path = "../output/製品別売上レポート.xlsx"

def convert_reiwa_date(text):
    match = re.match(r"令和(\d+)年(\d+)月(\d+)日", text)

    reiwa_year = int(match.group(1))
    month = int(match.group(2))
    day = int(match.group(3))

    year = reiwa_year + 2018

    return pd.Timestamp(year=year, month=month, day=day)

a_files = [
    "../input/発注書A①.xlsx",
    "../input/発注書A②.xlsx",
]
b_files = [
    "../input/発注書B①.csv",
    "../input/発注書B②.csv",
    "../input/発注書B③.csv"
]

a_dfs = []
b_dfs = []

for file in a_files:
    a_df = pd.read_excel(file)
    a_df = a_df.rename(columns={
        "発注日": "order_date",
        "商品名": "product_name",
        "数量": "quantity"
    })
    a_dfs.append(a_df)
    
for file in b_files:
    b_df = pd.read_csv(file)
    b_df = b_df.rename(columns={
        "発注日付": "order_date",
        "製品名": "product_name",
        "個数": "quantity"
    })
    b_df["order_date"] = b_df["order_date"].apply(convert_reiwa_date)

    b_dfs.append(b_df)

df_all = pd.concat(a_dfs + b_dfs,ignore_index=True)

df_summary = (
    df_all
    .groupby(["order_date", "product_name"], as_index=False)["quantity"]
    .sum()
)

wb = load_workbook(report_path)
ws = wb.active

start_row = 2
template_end_row = 21

# 通常行と最終行の書式を保存
normal_styles = []
last_styles = []

for col in range(1, 4):
    normal_cell = ws.cell(row=2, column=col)
    last_cell = ws.cell(row=21, column=col)

    normal_styles.append({
        "font": copy(normal_cell.font),
        "fill": copy(normal_cell.fill),
        "border": copy(normal_cell.border),
        "alignment": copy(normal_cell.alignment),
        "number_format": normal_cell.number_format,
    })

    last_styles.append({
        "font": copy(last_cell.font),
        "fill": copy(last_cell.fill),
        "border": copy(last_cell.border),
        "alignment": copy(last_cell.alignment),
        "number_format": last_cell.number_format,
    })

last_data_row = start_row + len(df_summary) - 1

for row_index, row in enumerate(
    df_summary.itertuples(index=False),
    start=start_row
):
    values = [
        row.order_date,
        row.product_name,
        row.quantity,
    ]

    for col_index, value in enumerate(values, start=1):
        cell = ws.cell(
            row=row_index,
            column=col_index,
            value=value
        )

        # 通常行の書式
        style = normal_styles[col_index - 1]

        cell.font = copy(style["font"])
        cell.fill = copy(style["fill"])
        cell.border = copy(style["border"])
        cell.alignment = copy(style["alignment"])
        cell.number_format = style["number_format"]

    # 日付だけ表示形式を指定
    ws.cell(
        row=row_index,
        column=1
    ).number_format = "yyyy/mm/dd"

for col_index in range(1, 4):
    cell = ws.cell(
        row=last_data_row,
        column=col_index
    )

    style = last_styles[col_index - 1]

    cell.border = copy(style["border"])

wb.save(output_path)
```
