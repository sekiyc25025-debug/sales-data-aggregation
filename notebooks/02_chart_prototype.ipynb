```python
import pandas as pd
from openpyxl import load_workbook
from openpyxl.chart.layout import Layout, ManualLayout

report_path = "../output/製品別売上レポート.xlsx"
test_output_path = "../output/製品別売上レポート_Ver2試作.xlsx"

df_report = pd.read_excel(report_path)

df_report.head()

df_report = df_report.rename(columns={
    "年月日": "order_date",
    "商品名": "product_name",
    "売上数量計": "quantity"
})

df_chart = df_report.pivot(
    index="order_date",
    columns="product_name",
    values="quantity"
).fillna(0)

wb = load_workbook(report_path)

ws_report = wb.active

if "グラフ用データ" in wb.sheetnames:
    del wb["グラフ用データ"]

ws_data = wb.create_sheet("グラフ用データ")

ws_data.cell(row=1, column=1, value="年月日")

for col_index, product_name in enumerate(df_chart.columns, start=2):
    ws_data.cell(
        row=1,
        column=col_index,
        value=product_name
    )

for row_index, (date, row) in enumerate(df_chart.iterrows(), start=2):
    ws_data.cell(
        row=row_index,
        column=1,
        value=date
    )

    ws_data.cell(
        row=row_index,
        column=1
    ).number_format = "yyyy/mm/dd"

    for col_index, value in enumerate(row, start=2):
        ws_data.cell(
            row=row_index,
            column=col_index,
            value=value
        )

from openpyxl.chart import LineChart, Reference

chart = LineChart()

chart.title = "商品別 日別売上数量"

# 軸タイトル
chart.x_axis.title = "年月日"
chart.y_axis.title = "売上数量"

# 軸そのものを表示
chart.x_axis.delete = False
chart.y_axis.delete = False

# 目盛ラベルの位置
chart.x_axis.tickLblPos = "low"
chart.y_axis.tickLblPos = "nextTo"

# グラフに使う数量データ
data = Reference(
    ws_data,
    min_col=2,
    max_col=1 + len(df_chart.columns),
    min_row=1,
    max_row=1 + len(df_chart)
)

# 横軸に使う日付
dates = Reference(
    ws_data,
    min_col=1,
    min_row=2,
    max_row=1 + len(df_chart)
)

# データを登録
chart.add_data(data, titles_from_data=True)
chart.set_categories(dates)

# 折れ線を直線にする
for series in chart.ser:
    series.smooth = False

# 凡例を右側へ
chart.legend.position = "r"

# プロット領域の調整
chart.layout = Layout(
    manualLayout=ManualLayout(
        x=-0.05,
        y=0,
        w=0.7,
        h=0.8
    )
)

chart.width = 18
chart.height = 10

ws_report.add_chart(chart, "E2")

wb.save(test_output_path)
```

```python
import pandas as pd
from openpyxl import load_workbook

report_path = "../output/製品別売上レポート.xlsx"
test_output_path = "../output/製品別売上レポート_Ver2試作.xlsx"

df_report = pd.read_excel(report_path)

df_report.head()

df_report = df_report.rename(columns={
    "年月日": "order_date",
    "商品名": "product_name",
    "売上数量計": "quantity"
})

df_chart = df_report.pivot(
    index="order_date",
    columns="product_name",
    values="quantity"
).fillna(0)

wb = load_workbook(report_path)

ws_report = wb.active

if "グラフ用データ" in wb.sheetnames:
    del wb["グラフ用データ"]

ws_data = wb.create_sheet("グラフ用データ")

ws_data.cell(row=1, column=1, value="年月日")

for col_index, product_name in enumerate(df_chart.columns, start=2):
    ws_data.cell(
        row=1,
        column=col_index,
        value=product_name
    )

for row_index, (date, row) in enumerate(df_chart.iterrows(), start=2):
    ws_data.cell(
        row=row_index,
        column=1,
        value=date
    )

    ws_data.cell(
        row=row_index,
        column=1
    ).number_format = "yyyy/mm/dd"

    for col_index, value in enumerate(row, start=2):
        ws_data.cell(
            row=row_index,
            column=col_index,
            value=value
        )

from openpyxl.chart import ScatterChart, Reference, Series
from openpyxl.chart.marker import Marker
from openpyxl.chart.axis import ChartLines
from openpyxl.chart.layout import Layout, ManualLayout
from openpyxl.utils.datetime import to_excel

chart = ScatterChart()

chart.title = "商品別 日別売上数量"

chart.x_axis.title = "年月日"
chart.y_axis.title = "売上数量"

chart.x_axis.delete = False
chart.y_axis.delete = False

chart.x_axis.tickLblPos = "low"
chart.y_axis.tickLblPos = "nextTo"

# X軸の表示間隔：2日
chart.x_axis.majorUnit = 2

# X軸の表示範囲を実データに合わせる
first_date = df_chart.index.min()
#last_date = df_chart.index.max()

chart.x_axis.scaling.min = to_excel(
    first_date.to_pydatetime()
)
#chart.x_axis.scaling.max = to_excel(
#    last_date.to_pydatetime()
#)

# X軸の日付表示
chart.x_axis.numFmt = "m/d"

# 折れ線＋プロット点
chart.scatterStyle = "lineMarker"

xvalues = Reference(
    ws_data,
    min_col=1,
    min_row=2,
    max_row=1 + len(df_chart)
)

for col in range(2, 2 + len(df_chart.columns)):
    values = Reference(
        ws_data,
        min_col=col,
        min_row=1,
        max_row=1 + len(df_chart)
    )

    series = Series(
        values,
        xvalues=xvalues,
        title_from_data=True
    )

    series.smooth = False
    series.marker = Marker(
        symbol="circle",
        size=5
    )

    chart.series.append(series)

# 縦グリッド線
chart.x_axis.majorGridlines = ChartLines()

chart.legend.position = "r"

chart.layout = Layout(
    manualLayout=ManualLayout(
        x=-0.05,
        y=0,
        w=0.7,
        h=0.8
    )
)

chart.width = 18
chart.height = 10

ws_report.add_chart(chart, "E2")

wb.save(test_output_path)
```
